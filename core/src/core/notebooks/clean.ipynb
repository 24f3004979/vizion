````python
# Foundational Data cleaning pipeline with pandas route

UNIQUE_THRESHOLD = 90

indexing_elem = [] # indexing elements

# Filter 1 : Indexing columns separation
for col in df_test.columns:
    elems = len(df_test[col].unique())
    total_elem = len(df_test[col])
    unique_rate = (elems/total_elem) * 100
    if unique_rate > UNIQUE_THRESHOLD:
        indexing_elem.append(col)

print("Indexing Elements",indexing_elem)

df = df_test.drop(indexing_elem, axis=1)

string_columns = [ col for col in df.columns if df[col].dtype == 'string' ]
numeric_columns = [ col for col in df.columns if df[col].dtype in 'int,float'.split(',') ]

# Foundational String columns filteration with pd tools

for col in string_columns:
    element = df[col]
    df[col] = element.str.strip()
    df[col] = element.str.lower()

'''
For numerical dtypes user would need to enter
1. Make column absolute with just integer
2. clip into a range
3. List out outliers
4. What to fill into null values
'''

print(f"String Columns : {string_columns}")
print(f"Numerical Columns : {numeric_columns}")

df
````

````python
# 1. Null values reporting execution
'''
Loading for checking null and nan values into the given dataset.
Edge cases to fight
Foundational string based checkig power
    Load up for normalizing during check with string protection
    loading up everyentry with apply function to check with null and Nan values

Working pipeline
Clean with pandas  - Run custom pipeline with string protection - get final report with string based elements
'''
import pandas as pd

NULL_REPORT = {}

# Before running custom pipeline for string protection remove the nans first
df_test_copy = df_test.copy()
df_test = df_test.dropna()

# Initial initiating sequence must required
for col in df_test:
    NULL_REPORT[col] = 0

def null_protection(entry:str):
    '''
    checks entry for null and none values
    input:entry
    returns:boolean
    '''
    null_group = ['null', 'none', 'nan']  # could have more

    normalized = entry.lower()
    if (normalized in null_group):
        return True
        
    return False

# Series level operation with null protections
def null_check(series:pd.Series):
    '''
    Total null count with string check
    input:Series
    return:int
    '''
    count = 0
    for element in series:
        check = null_protection(element)
        if check:
            count += 1
    return count

# Loading Null reporting valus for given dataset
string_columns = ['plan_type', 'device']

for col in string_columns:
    print("target", col)
    series = df_test[col]
    count = null_check(series)
    NULL_REPORT[col]+=count

print(NULL_REPORT)
````
